```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/cleaned_netflix.csv")
```

```python
df.head()
```

```python
print("Dataset Shape", df.shape)
```

```python
print("Earliest Release Year:", df["release_year"].min())
print("Latest Release Year:", df["release_year"].max())
```

```python
df["release_year"].describe()
```

```python
yearly_content = df["release_year"].value_counts().sort_index()
print(yearly_content)
```

```python
top_release_years = yearly_content.sort_values(ascending=False).head(10)

print(top_release_years)
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    yearly_content.index,
    yearly_content.values,
    marker="o"
)

plt.title("Netflix Content Production by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.grid(True)

plt.show()
```

```python
latest_year = df["release_year"].max()

recent_years = yearly_content[
    yearly_content.index >= latest_year - 10
]

print(recent_years)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=recent_years.index,
    y=recent_years.values
)

plt.title("Netflix Content Released in Recent Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)

plt.show()
```

```python
peak_year = yearly_content.idxmax()
peak_count = yearly_content.max()

print("Peak Release Year:", peak_year)
print("Number of Titles:", peak_count)
```

```python
earliest_year = yearly_content.index.min()
latest_year = yearly_content.index.max()

print("Earliest Year:", earliest_year)
print("Titles in Earliest Year:", yearly_content.loc[earliest_year])

print("Latest Year:", latest_year)
print("Titles in Latest Year:", yearly_content.loc[latest_year])
```

Conclusion:

The release-year analysis shows that the Netflix dataset contains titles
released between 1925 and 2021. Content production increased considerably
in the years leading up to 2018, which recorded the highest number of titles
at 1,146.

After reaching the peak in 2018, the number of titles decreased to 1,030
in 2019, 953 in 2020, and 592 in 2021. Overall, the dataset shows a strong
growth pattern toward the late 2010s, followed by a decline in the number
of titles in the final years represented in the dataset.
