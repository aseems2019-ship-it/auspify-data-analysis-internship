```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/cleaned_netflix.csv")
df.head()
```

```python
print("Dataset Shape", df.shape)
```

```python
df.info()
```

```python
content_counts = df["type"].value_counts()
print(content_counts)
```

```python
movie_count = (df["type"] == "Movie").sum()
tv_show_count = (df["type"] == "TV Show").sum()
print("Total Movies:", movie_count)
print("Total TV Shows:", tv_show_count)
```

```python
content_percentage = df["type"].value_counts(normalize=True) * 100
print(content_percentage.round(2))
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x="type")
plt.title("Distribution of Movies and TV Shows on Netflix")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")
plt.show()
```

```python
plt.figure(figsize=(7,7))
df["type"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Movies vs TV shows on Netflix")
plt.ylabel("")
plt.show()
```

```python
dominant_type = df["type"].value_counts().idxmax()
dominant_count = df["type"].value_counts().max()

print("Most common content type:", dominant_type)
print("Number of titles:", dominant_count)
```
